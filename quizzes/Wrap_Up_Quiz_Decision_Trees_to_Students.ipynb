````python
import pandas as pd
import sklearn as skl

ames_housing = pd.read_csv("https://raw.githubusercontent.com/bilals/scikit-learn-mooc/main/datasets/ames_housing_no_missing.csv")
target_name = "SalePrice"
data = ames_housing.drop(columns=target_name)
target = ames_housing[target_name]
numerical_features = [
    "LotFrontage", "LotArea", "MasVnrArea", "BsmtFinSF1", "BsmtFinSF2",
    "BsmtUnfSF", "TotalBsmtSF", "1stFlrSF", "2ndFlrSF", "LowQualFinSF",
    "GrLivArea", "BedroomAbvGr", "KitchenAbvGr", "TotRmsAbvGrd", "Fireplaces",
    "GarageCars", "GarageArea", "WoodDeckSF", "OpenPorchSF", "EnclosedPorch",
    "3SsnPorch", "ScreenPorch", "PoolArea", "MiscVal",
]

data_numerical = data[numerical_features]
````

````python
linear_model = skl.pipeline.Pipeline([
    ('scaler', skl.preprocessing.StandardScaler()),
    ('linear_regression', skl.linear_model.LinearRegression())
])

tree_model = skl.tree.DecisionTreeRegressor(random_state=0)


cv_results_linear = skl.model_selection.cross_validate(linear_model, data_numerical, target, cv=10)
cv_results_tree = skl.model_selection.cross_validate(tree_model, data_numerical, target, cv=10)

linear_scores = cv_results_linear['test_score']
tree_scores = cv_results_tree['test_score']

count_linear_better = sum(linear_scores > tree_scores for linear_scores, tree_scores in zip(linear_scores, tree_scores))
print(f"Number of folds where linear model is better: {count_linear_better}")
print(f"Linear scores: {linear_scores}")
print(f"Tree scores: {tree_scores}")
````

Question 1 : C)


````python
param_grid = {'max_depth': range(1, 16)}
tree_model = skl.tree.DecisionTreeRegressor(random_state=0)
grid_search = skl.model_selection.GridSearchCV(tree_model, param_grid, cv=10)
grid_search.fit(data_numerical, target)

print("Best depth:", grid_search.best_params_['max_depth'])
print("Score  :", grid_search.best_score_)
````

````python
# Tuned tree
param_grid = {'max_depth': range(1, 16)}
tree_model = skl.tree.DecisionTreeRegressor(random_state=2)
tuned_tree = skl.model_selection.GridSearchCV(tree_model, param_grid, cv=10)
# Linear model
linear_model = skl.pipeline.Pipeline([
    ('scaler', skl.preprocessing.StandardScaler()),
    ('linear_regression', skl.linear_model.LinearRegression())
])

cv_results_tuned_tree = skl.model_selection.cross_validate(tuned_tree, data_numerical, target, cv=10)
cv_results_linear = skl.model_selection.cross_validate(linear_model, data_numerical, target, cv=10)

tuned_tree_scores = cv_results_tuned_tree['test_score']
linear_scores = cv_results_linear['test_score']
count_tuned_better = sum(t > l for t, l in zip(tuned_tree_scores, linear_scores))
print(f"Folds where tuned tree is better: {count_tuned_better}/10")
````
